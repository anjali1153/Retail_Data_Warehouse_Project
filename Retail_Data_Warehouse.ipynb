```python
import pandas as pd
import numpy as np
from sqlalchemy import create_engine
```

```python
df = pd.read_csv(r"D:\Document 19-05-2026\Retail_Sales_Data_Warehouse\data\Sample - Superstore.csv", encoding="latin1")
```

```python
df.columns = (
    df.columns
      .str.lower()
      .str.replace(" ", "_")
      .str.replace("-", "_")
)

df.head()
```

```python
df.isnull().sum()
```

```python
df['order_date'] = pd.to_datetime(df['order_date'])
df['ship_date'] = pd.to_datetime(df['ship_date'])
```

```python
df.dtypes
```

```python
df['year'] = df['order_date'].dt.year
```

```python
df['month'] = df['order_date'].dt.month
```

```python
df['quarter'] = df['order_date'].dt.quarter
```

```python
df[['order_date','year','month','quarter']].head()
```

```python
engine = create_engine(
    "mysql+pymysql://root:Angel*1153@localhost/retail_dw"
)
```

```python
df.to_sql(
    "stg_orders",
    con=engine,
    if_exists="replace",
    index=False
)
```

```python
query = """
SELECT *
FROM stg_orders
LIMIT 5
"""

pd.read_sql(query, engine)
```

```python
# Total Sales

df['sales'].sum()
```

```python
# Total Profit

df['profit'].sum()
```

```python
# Region wise sales

df.groupby('region')['sales'].sum()
```

```python
# Category wise sales`

df.groupby('category')['sales'].sum()
```

```python
# Top 10 customers

top_customers = (
    df.groupby('customer_name')['sales']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

top_customers
```

```python
# Customer Dimension

dim_customer = df[
    ['customer_id',
     'customer_name',
     'segment']
].drop_duplicates()
```

```python
# Load to MySQL

dim_customer.to_sql(
    'dim_customer',
    engine,
    if_exists='replace',
    index=False
)
```

```python
# Product Dimension

dim_product = df[
    ['product_id',
     'product_name',
     'category',
     'sub_category']
].drop_duplicates()
```

```python
# Load to MySQL

dim_product.to_sql(
    'dim_product',
    engine,
    if_exists='replace',
    index=False
)
```

```python
# Region Dimension

dim_region = df[
    ['region',
     'state',
     'city']
].drop_duplicates()
```

```python
# Load to MySQL

dim_region.to_sql(
    'dim_region',
    engine,
    if_exists='replace',
    index=False
)
```

```python
# Date Dimension

dim_date = df[
    ['order_date',
     'year',
     'month',
     'quarter']
].drop_duplicates()
```

```python
# Load to MySQL

dim_date.to_sql(
    'dim_date',
    engine,
    if_exists='replace',
    index=False
)
```

```python
# Create Fact Table

fact_sales = df[
[
'order_id',
'customer_id',
'product_id',
'order_date',
'region',
'sales',
'quantity',
'discount',
'profit'
]
]
```

```python
fact_sales.to_sql(
    'fact_sales',
    engine,
    if_exists='replace',
    index=False
)
```
